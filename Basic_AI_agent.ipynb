```python
"""
AI-Augmented Workflow - Basic Agent
Self-Learning Activity

This is a simple rule-based personal assistant.
It accepts user commands and performs basic tasks.
"""


def calculate(expression):
    """Safely evaluate a basic mathematical expression."""
    try:
        allowed_characters = "0123456789+-*/(). "

        if all(character in allowed_characters for character in expression):
            result = eval(expression)
            return f"Result: {result}"

        return "Please enter a valid mathematical expression."

    except ZeroDivisionError:
        return "Error: Cannot divide by zero."

    except Exception:
        return "Sorry, I could not calculate that."


def get_response(command):
    """Generate a response based on the user's command."""

    command = command.lower().strip()

    if command in ["hello", "hi", "hey"]:
        return "Hello! I am your Basic AI Assistant."

    elif "your name" in command:
        return "My name is Basic AI Assistant."

    elif "help" in command:
        return (
            "I can help you with:\n"
            "1. Greetings\n"
            "2. Simple calculations\n"
            "3. Basic information\n"
            "4. Exit the program"
        )

    elif "study" in command:
        return "Tip: Break your study session into small focused tasks."

    elif "python" in command:
        return "Python is a popular programming language used in AI and data science."

    elif "ai" in command:
        return "AI stands for Artificial Intelligence. It enables machines to perform intelligent tasks."

    elif command.startswith("calculate"):
        expression = command.replace("calculate", "", 1).strip()

        if expression:
            return calculate(expression)

        return "Please provide a mathematical expression."

    elif command in ["bye", "exit", "quit"]:
        return "Goodbye! Have a productive day."

    else:
        return (
            "I don't understand that command yet. "
            "Type 'help' to see what I can do."
        )


def run_agent():
    """Start and continuously run the AI agent."""

    print("=" * 45)
    print("       BASIC AI PERSONAL ASSISTANT")
    print("=" * 45)
    print("Type 'help' to see available commands.")
    print("Type 'exit' to close the assistant.\n")

    while True:
        user_input = input("You: ")

        response = get_response(user_input)

        print("Agent:", response)

        if user_input.lower().strip() in ["bye", "exit", "quit"]:
            break


if __name__ == "__main__":
    run_agent()
```
